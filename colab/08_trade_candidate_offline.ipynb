```python
# 1. Fresh, code-only, no-Drive runtime.
import os,subprocess,sys,hashlib,json
from pathlib import Path
if any(Path(p).exists() for p in ('/content/drive/MyDrive','/content/gdrive/MyDrive','/gdrive/MyDrive')):
    raise RuntimeError('Drive is mounted; delete runtime and use clean offline Colab.')
if any(os.environ.get(k) for k in ('GOOGLE_APPLICATION_CREDENTIALS','GOOGLE_OAUTH_ACCESS_TOKEN','GOOGLE_DRIVE_TOKEN')):
    raise RuntimeError('Credentials detected; fail closed.')
repo=Path('/content/hypolab-strategy-v1-code')
if repo.exists():
    raise RuntimeError('Fresh Colab runtime required; do not reuse code checkout.')
subprocess.run(['git','clone','--depth','1','https://github.com/DuStER1921/CryptoHypothesisLab.git',str(repo)],check=True,stdout=subprocess.DEVNULL)
commit=subprocess.check_output(['git','rev-parse','HEAD'],cwd=repo,text=True).strip()
sys.path.insert(0,str(repo/'src'))
from hypolab.strategy_v1 import make_strategy_plan,run_strategy_v1,freeze_token_sha256
from hypolab.micro_bundle import MAX_ARCHIVE_BYTES
from google.colab import files
print('FROZEN STRATEGY CODE COMMIT:',commit)
print('NO DRIVE; ORIGINAL RESEARCH DATA INACCESSIBLE')
```

```python
# 2. Upload ONE private MICRO ZIP, verify SHA256 BEFORE any outcomes.
uploaded=files.upload()
if len(uploaded)!=1:raise RuntimeError('Exactly one private MICRO ZIP required')
name,raw=next(iter(uploaded.items()))
if not name.startswith('HYPOLAB_MICRO_DEV_') or not name.endswith('.zip') or not 0<len(raw)<=MAX_ARCHIVE_BYTES:
    raise RuntimeError('Unexpected file name/size')
expected=input('Exact SHA-256 of this MICRO ZIP: ').strip().lower()
if hashlib.sha256(raw).hexdigest()!=expected:
    raise RuntimeError('MICRO ZIP SHA-256 mismatch; fail closed')
private=Path('/content/hypolab-strategy-v1-private')
private.mkdir(mode=0o700,exist_ok=False)
packet=private/'micro-input.zip'
with packet.open('xb') as f:
    os.chmod(packet,0o600)
    f.write(raw)
del uploaded,raw
plan=make_strategy_plan(packet)
fields=json.loads(plan)
month=fields['month']
print('Verified month:',month,'kind:',fields['kind'])
print('FROZEN STRATEGY V1 TOKEN SHA256:',fields['freeze_token_sha256'])
if month=='2024-08':
    owner_token=input('Paste EXACT saved July Strategy V1 freeze token: ').strip().lower()
    if owner_token!=fields['freeze_token_sha256']:
        raise RuntimeError('FROZEN STRATEGY CODE DRIFT; DO NOT CONTINUE')
```

```python
# 3. Save exact strategy research PLAN before reading labels.
plan_path=private/'strategy-v1-frozen-plan.json'
with plan_path.open('xb') as f:f.write(plan)
print('PLAN SHA-256:',hashlib.sha256(plan).hexdigest())
print('SAVE TOKEN FOR AUGUST:',fields['freeze_token_sha256'])
files.download(str(plan_path))
```

```python
# 4. Run ONE fixed strategy, same-time opposite-side control, cost stress.
report=run_strategy_v1(packet,plan)
p=report['primary'];b=report['opposite_side_control']
print('MODEL STATUS:',report['model_status'])
print('Signals:',report['signal_count_before_execution_constraints'],'Trades:',p['trades'],'Days:',p['active_utc_days'])
print('Net mean bps (with ASSUMED costs):',p['net_mean_bps'])
print('Opposite-side mean bps:',b['net_mean_bps'])
print('Day-cluster 95% descriptive range:',p['day_cluster_bootstrap_95pct_mean_net_bps'])
print('Stressed slippage:',report['stress'])
print('Unknown/excluded:',report['skips'])
for name,passed in report['gates'].items():
    print(('PASS' if passed else 'FAIL'), name)
print('REAL EXECUTABLE PROFITABILITY: NOT PROVEN')
```

```python
# 5. Fixed chart of every executed *modeled* trade. No event cherry-picking.
import matplotlib.pyplot as plt
pnl=[r['net_bps'] for r in report['trades']]
opp=[r['net_bps'] for r in report['opposite_control_trades']]
if pnl:
    a=[0.];b=[0.]
    for x,y in zip(pnl,opp):a.append(a[-1]+x);b.append(b[-1]+y)
    fig,ax=plt.subplots(figsize=(10,4))
    ax.plot(a,label='Predeclared strategy; ASSUMED cost')
    ax.plot(b,label='Same-time opposite-side hypothetical control')
    ax.set(xlabel='Trade sequence (not calendar time)',ylabel='Cumulative per-unit notional bps',
           title='DESCRIPTIVE ONLY — synthetic trade-price entry/exit, NOT executable P&L')
    ax.legend();ax.grid(True,alpha=.2);plt.show()
else:print('NO MODELED TRADES — valid negative result; do not tune thresholds.')
```
